```python
import numpy as np

from scAnalysis import (
    sc_io,
    preprocessing,
    quality_control,
    cell_cycle,
    batch_correction,
    dimensionality,
    clustering,
    trajectory,
    differential,
    enrichment,
    visualization,
    interactive_viz,
    imputation,
)

import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="statsmodels")
```

```python
local_h5ad_path = "/content/drive/MyDrive/resources/grn_benchmark/evaluation_data/MSCIC_sc.h5ad"
data = sc_io.read_h5ad(local_h5ad_path)
```

```python
data.var.index = sc_io._make_unique(data.var.index.values)
print(f"Loaded {data.n_obs} cells and {data.n_vars} genes from {local_h5ad_path}.")
```

```python
data
```

```python
data.var.index = sc_io._make_unique(data.var.index.values)
```

```python
preprocessing.calculate_qc_metrics(data, qc_vars=["MT-", "mt-"])
```

```python
quality_control.scrublet(data, verbose=True)
mask_singlets = ~data.obs['predicted_doublet'].astype(bool)
data = data[mask_singlets, :]
```

```python
data = preprocessing.filter_cells(data, min_genes=200, max_pct_mito=15.0)
data = preprocessing.filter_genes(data, min_cells=3)
```

If you intend to use this data to determine how genes interact with one another (their correlations and network structures), performing imputation is one of the biggest mistakes you can make.

Imputation algorithms (such as your WNID or industry-standard algorithms like MAGIC) fill in zero-value gene expressions based on the average of neighboring cells. This inherently creates false correlations between genes that are actually unrelated. Gene Regulatory Network (GRN) algorithms are then misled by these false correlations, resulting in the construction of inaccurate biological networks.

```python
preprocessing.normalize_total(data, target_sum=1e4)
preprocessing.log1p(data)
```

```python
preprocessing.highly_variable_genes(data, n_top_genes=2000)
```

```python
data.raw = data.copy()
```

```python
preprocessing.scale(data, max_value=10)
```

```python
dimensionality.run_pca(data, n_components=50)
```

```python
batch_correction.harmony_integrate(
    data,
    batch_key='donor_id',
    basis='X_pca',
    adjusted_basis='X_pca_harmony',
    verbose=True
)
```

```python
data.obsm['X_pca_unintegrated'] = data.obsm['X_pca']
data.obsm['X_pca'] = data.obsm['X_pca_harmony']

dimensionality.neighbors(data, n_neighbors=15, n_pcs=40)
dimensionality.run_umap(data, min_dist=0.3)
```

```python
clustering.cluster_leiden(data, resolution=0.5, key_added="leiden")
```

```python
print(data.obs['perturbation'].value_counts())
```

```python
differential.rank_genes_groups(data, groupby='leiden', method='wilcoxon', use_raw=True)
```

```python
first_cluster = sorted(data.obs['leiden'].unique())[0]
```

```python
cluster_markers = differential.get_marker_genes(data, group=first_cluster, pval_cutoff=0.05, lfc_cutoff=0.5)
print(f"Top 5 markers for cluster '{first_cluster}':{cluster_markers.head(5)}")
```

```python
try:
    interactive_viz.interactive_embedding(
        data,
        basis='X_umap',
        color='leiden',
        hover_data=['donor_id', 'cell_type'],
        title="Interactive UMAP - Leiden Clusters (MSCIC Dataset)",
        save_html="mscic_interactive_umap.html"
    )

    interactive_viz.interactive_3d_embedding(
        data,
        basis='X_pca',
        color='leiden',
        dimensions=[0, 1, 2],
        save_html="mscic_interactive_3d_pca.html"
    )
    if 'cluster_markers' in locals() and not cluster_markers.empty:
        top_genes = cluster_markers['names'].head(10).tolist()
        valid_genes = [g for g in top_genes if g in data.var.index]

        if valid_genes:
            interactive_viz.interactive_heatmap(
                data,
                var_names=valid_genes,
                groupby='leiden',
                use_raw=True,
                save_html="mscic_interactive_heatmap.html"
            )

except Exception as e:
    print(f"Skipped interactive plots. Please ensure 'plotly' is installed (pip install plotly): {e}")
```
